```python
import sys
sys.path.insert(0, '..')

import pandas as pd

equity_trs = pd.read_csv('../trade_data/equity_trs.csv')
bond_forward = pd.read_csv('../trade_data/bond_forward.csv')
bond_trs = pd.read_csv('../trade_data/bond_trs.csv')
equities = pd.read_csv('../trade_data/underlyings/equities.csv')
bonds = pd.read_csv('../trade_data/underlyings/bonds.csv')

print('Equity TRS:', len(equity_trs))
print('Bond Forward:', len(bond_forward))
print('Bond TRS:', len(bond_trs))
print('Total trades:', len(equity_trs) + len(bond_forward) + len(bond_trs))
```

```python
# Counterparty breakdown across all three instrument types
all_cptys = pd.concat([
    equity_trs[['trade_id', 'counterparty']],
    bond_forward[['trade_id', 'counterparty']],
    bond_trs[['trade_id', 'counterparty']],
])
all_cptys['counterparty'].value_counts()
```

```python
# JPY compo trades
jpy_trades = equities.loc[equities['currency'] == 'JPY', 'trade_id'].unique()
print('JPY-basket trade IDs:', jpy_trades)

print('Unique underlying names:', equities['isin'].nunique())
print('Unique bond types:', bonds['bond_id'].nunique())
```
